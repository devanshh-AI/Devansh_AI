```python
# Q2 — Student Grade Analyzer

def add_student(name, marks_list):
    average = sum(marks_list) / len(marks_list)

    if average >= 90:
        grade = "A"
    elif average >= 80:
        grade = "B"
    elif average >= 70:
        grade = "C"
    elif average >= 60:
        grade = "D"
    else:
        grade = "F"

    with open("students.txt", "a") as file:
        file.write(f"{name} | {average:.2f} | grade: {grade}\n")

    print(f"{name} added successfully!")


def show_all_students():
    print("\n=== All Students ===")
    with open("students.txt", "r") as file:
        lines = file.readlines()
        for line in lines:
            print(line.strip())


def topper_finder():
    best_avg  = 0
    best_name = ""

    with open("students.txt", "r") as file:
        lines = file.readlines()
        for line in lines:
            parts   = line.strip().split(" | ")
            name    = parts[0]
            average = float(parts[1])
            if average > best_avg:
                best_avg  = average
                best_name = name

    print(f"\n🏆 Topper: {best_name} with {best_avg}%")


add_student("Devansh", [85, 90, 92])
add_student("Rahul",   [70, 75, 68])
add_student("Priya",   [95, 88, 91])
add_student("John",    [60, 55, 58])
show_all_students()
topper_finder()
```

```python
# Q3 — Job Filter

def filter_jobs(jobs, location, min_salary):
    print("\n=== Matching Jobs ===")
    for job in jobs:
        job_title    = job["title"]
        job_location = job["location"]
        job_salary   = job["salary"]

        if job_location == location and job_salary >= min_salary:
            print(f"Title    : {job_title}")
            print(f"Location : {job_location}")
            print(f"Salary   : {job_salary}")
            print("--------")


jobs = [
    {"title": "AI Engineer",    "location": "Delhi",  "salary": 120000},
    {"title": "Python Dev",     "location": "Mumbai", "salary": 80000},
    {"title": "ML Engineer",    "location": "Delhi",  "salary": 95000},
    {"title": "Data Scientist", "location": "Pune",   "salary": 110000},
    {"title": "Java Dev",       "location": "Delhi",  "salary": 70000}
]
filter_jobs(jobs, "Delhi", 90000)
```

```python
# Q4 — Attendance Tracker

def attendance_report(employees):
    print("\n=== Attendance Report ===")
    for employee in employees:
        name       = employee["name"]
        present    = employee["present"]
        total      = employee["total"]
        attendance = (present / total) * 100

        if attendance >= 90:
            grade = "Excellent"
        elif attendance >= 75:
            grade = "Good"
        else:
            grade = "Poor"

        print(f"{name} | {attendance:.1f}% | {grade}")


employees_attendance = [
    {"name": "Devansh", "present": 24, "total": 26},
    {"name": "Rahul",   "present": 18, "total": 26},
    {"name": "Priya",   "present": 25, "total": 26},
    {"name": "John",    "present": 15, "total": 26}
]
attendance_report(employees_attendance)
```

```python
# Q5 — Top Performer Finder

def find_top_performer(employees):
    print("\n=== Top Performer ===")
    max_score = 0
    top_name  = ""

    for employee in employees:
        name  = employee["name"]
        score = employee["score"]

        if score > max_score:
            max_score = score
            top_name  = name

    print(f"🏆 Top Performer: {top_name} ({max_score})")


employees_score = [
    {"name": "Devansh", "score": 85},
    {"name": "Rahul",   "score": 92},
    {"name": "Priya",   "score": 78},
    {"name": "John",    "score": 95},
    {"name": "Sara",    "score": 88}
]
find_top_performer(employees_score)
```

```python
# Q6 — Resume Saver

def save_resumes(candidates):
    print("\n=== Saving Resumes ===")
    with open("resumes.txt", "a") as file:
        for candidate in candidates:
            name = candidate["name"]
            role = candidate["role"]
            exp  = candidate["exp"]
            file.write(f"{name} | {role} | {exp} years\n")
            print(f"{name} saved!")


def read_resumes():
    print("\n=== All Resumes ===")
    with open("resumes.txt", "r") as file:
        lines = file.readlines()
        for line in lines:
            print(line.strip())


candidates = [
    {"name": "Devansh", "role": "AI Engineer",   "exp": 3},
    {"name": "Rahul",   "role": "Python Dev",     "exp": 2},
    {"name": "Priya",   "role": "Data Scientist", "exp": 4}
]
save_resumes(candidates)
read_resumes()
```
