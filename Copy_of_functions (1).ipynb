```python
# Simple Function — No Parameter
def greet():
    print("Hello! I am learning Python!")

greet()
```

```python
# Function with 1 Parameter
def greet(name):
    print(f"Hello {name}! I am learning Python!")

greet("Devansh")
greet("Rahul")
greet("Priya")
```

```python
# Function with 2 Parameters
def greet(name, city):
    print(f"Hello {name}!")
    print(f"I am from {city}!")
    print("---")

greet("Devansh", "Kanpur")
greet("Rahul", "Mumbai")
greet("Priya", "Delhi")
```

```python
# Return — Function Answer Wapas Bheje
def calculate_bonus(salary):
    bonus = salary * 0.15
    return bonus

my_bonus = calculate_bonus(62000)
print(my_bonus)

# Multiple calls
print(calculate_bonus(80000))
print(calculate_bonus(95000))
```

```python
# CTC Calculator
def CTC_Calculator(salary):
    bonus = salary * 0.15
    ctc = bonus + salary
    return ctc

print(CTC_Calculator(62000))
print(CTC_Calculator(80000))
print(CTC_Calculator(95000))
```

```python
# Candidate Information
def candidate_info(name, role, salary):
    bonus = salary * 0.10

    print("=== Candidate Information ===")
    print(f"Name  : {name}")
    print(f"Role  : {role}")
    print(f"Salary: ${salary}")
    print(f"Bonus : ${bonus}")
    print("=" * 30)

candidate_info("Devansh", "AI Engineer", 62000)
candidate_info("Rahul", "Python Developer", 80000)
candidate_info("Priya", "Data Scientist", 95000)
```

```python
# Candidate Checker
def check_candidate(name, experience, score):
    if experience >= 3 and score >= 70:
        print(f"{name} is Selected! ✅")
    else:
        print(f"{name} is Rejected ❌")

check_candidate("Devansh", 3, 80)
check_candidate("Rahul", 1, 90)
check_candidate("Priya", 5, 65)
```

```python
# Salary Checker with Return
def salary_check(name, salary):
    if salary >= 100000:
        level = "Senior"
    elif salary >= 50000:
        level = "Mid Level"
    else:
        level = "Junior"
    return level

print(salary_check("Devansh", 62000))
print(salary_check("Rahul", 120000))
print(salary_check("Priya", 40000))
```

```python
# Complete Offer Letter Function
def offer_letter(name, role, company, salary):
    bonus = salary * 0.15
    total_ctc = salary + bonus

    print("=" * 40)
    print("        OFFER LETTER")
    print("=" * 40)
    print(f"Dear {name},")
    print(f"Role    : {role}")
    print(f"Company : {company}")
    print(f"Salary  : ${salary}")
    print(f"Bonus   : ${bonus}")
    print(f"Total   : ${total_ctc}")
    print("=" * 40)
    print("Congratulations! Welcome aboard.")
    print()

offer_letter("Devansh", "AI Engineer", "TechCorp", 120000)
offer_letter("Rahul", "Python Dev", "Google", 150000)
offer_letter("Priya", "Data Scientist", "Amazon", 130000)
```
