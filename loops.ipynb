```python
# For Loop — List
fruits = ["Mango", "Banana", "Apple"]
for fruit in fruits:
    print(f"I love {fruit}!")
```

```python
# Q1 — Cities
cities = ["Delhi", "Mumbai", "Kanpur", "Bangalore", "Chennai"]
for city in cities:
    print(f"I want to visit {city}!")
```

```python
# Q2 — Salary Slab
salaries = [45000, 82000, 35000, 95000, 61000]
for salary in salaries:
    if salary < 50000:
        print(f"{salary} → Junior Level")
    elif salary <= 90000:
        print(f"{salary} → Mid Level")
    else:
        print(f"{salary} → Senior Level")
```

```python
# Q3 — Even Numbers
for i in range(2, 11, 2):
    print(i)
```

```python
# Q4 — Countdown
for i in range(10, 0, -1):
    print(f"{i}.....")
print("Happy New Year! 🎉")
```

```python
# Q5 — Candidates + Scores
candidates = ["Devansh", "Rahul", "Priya", "John"]
scores = [85, 45, 72, 90]

for i in range(len(candidates)):
    if scores[i] >= 70:
        print(f"{candidates[i]} scored {scores[i]} → Selected ✅")
    else:
        print(f"{candidates[i]} scored {scores[i]} → Rejected ❌")
```

```python
# Q6 — Companies
companies = ["Google", "Amazon", "Microsoft"]
packages = [150000, 130000, 140000]

for i in range(len(companies)):
    print(f"{i+1}. {companies[i]} → ${packages[i]}")
```

```python
# Q7 — Skills Enumerate
skills = ["Python", "SQL", "AI", "ML", "RAG"]

for index, skill in enumerate(skills):
    print(f"Skill {index+1}: {skill}")
```

```python
# Q8 — Movies Enumerate
movies = ["Tere Nam", "Dabbang", "Ready", "Bodyguard", "Sultan"]

for index, movie in enumerate(movies):
    print(f"Rank {index+1}: {movie} 🎬")
```

```python
# While Loop Basic
count = 1
while count <= 5:
    print(f"Count: {count}")
    count += 1
```

```python
# While Countdown
count = 5
while count > 0:
    print(f"{count}...")
    count -= 1
print("Blast off! 🚀")
```

```python
# Break — Pehla 20+ number
numbers = [10, 25, 3, 47, 8, 15]
for num in numbers:
    if num > 20:
        print(f"Found: {num}")
        break
```

```python
# Continue — AI Skills
skills = ["Python", "Java", "AI", "Ruby", "ML", "PHP"]
ai_skills = ["Python", "AI", "ML"]

for skill in skills:
    if skill not in ai_skills:
        continue
    print(f"AI Skill: {skill} ✅")
```

```python
# Loop + Function combo!
def check_candidate(name, score):
    if score >= 70:
        print(f"{name} → Selected! ✅")
    else:
        print(f"{name} → Rejected ❌")

candidates = ["Devansh", "Rahul", "Priya", "John"]
scores = [85, 45, 72, 90]

for i in range(len(candidates)):
    check_candidate(candidates[i], scores[i])
```

```python
# Dictionary Loop
candidate = {
    "name"   : "Devansh",
    "city"   : "Kanpur",
    "role"   : "AI Engineer",
    "salary" : 62000
}

for key, value in candidate.items():
    print(f"{key} : {value}")
```

```python
# Q9 — Salary Growth
salary = 50000
year = 1
while salary < 100000:
    print(f"Year {year}: ${salary}")
    salary += salary * 0.10
    year += 1
print(f"Target reached: ${salary:.0f} 🎯")
```

```python
# Q10 — Odd Numbers
count = 1
while count <= 10:
    if count % 2 != 0:
        print(count)
    count += 1
```

```python
# Q11 — Break
candidates = ["Rahul", "Priya", "Devansh", "John", "Sara"]
scores     = [45, 92, 78, 55, 88]

for i in range(len(candidates)):
    if scores[i] >= 90:
        print(f"Top candidate: {candidates[i]} ({scores[i]}) 🎯")
        break
```

```python
# Q12 — Break Divisible
for i in range(1, 51):
    if i % 7 == 0:
        print(f"First number divisible by 7: {i}")
        break
```

```python
# Q13 — Continue
numbers = [12, 7, 45, 3, 89, 23, 56]
for number in numbers:
    if number <= 10:
        continue
    print(f"{number} ✅")
```

```python
# Q14 — Continue Empty
names = ["Devansh", "", "Rahul", "", "Priya"]
for name in names:
    if name == "":
        continue
    print(f"Candidate: {name} ✅")
```

```python
# Q15 — Nested Loop
interviewers = ["HR", "Technical", "Manager"]
candidates   = ["Devansh", "Rahul"]

for interviewer in interviewers:
    for candidate in candidates:
        print(f"{interviewer} interviewing {candidate}")
    print("---")
```

```python
# Q16 — Multiplication Table
for i in range(2, 4):
    for j in range(1, 11):
        print(f"{i} x {j} = {i*j}")
    print("---")
```
