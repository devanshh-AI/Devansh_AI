```python
# Job Postings CSV
with open("job_posting.csv", "w") as file:
    file.write("JobTitle,Location,MinSalary,MaxSalary,Remote\n")
    file.write("AI Engineer,Delhi,100000,150000,Yes\n")
    file.write("Python Developer,Mumbai,80000,120000,No\n")
    file.write("Java Dev,Noida,50000,89000,Yes\n")
    file.write("Fullstack Dev,Pune,60000,80000,No\n")

print("job_posting.csv created! ✅")
```

```python
# CSV Line by Line
with open("job_posting.csv", "r") as file:
    for line in file:
        print(line.strip())
```

```python
import csv

# csv.reader — Index se
with open("job_posting.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
import csv

# Header Skip karo
with open("job_posting.csv", "r") as f:
    reader = csv.reader(f)
    next(reader)
    for row in reader:
        print(f"Role: {row[0]} | Location: {row[1]}")
```

```python
import csv

# DictReader — Name se
with open("job_posting.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"Role          : {row['JobTitle']}")
        print(f"City          : {row['Location']}")
        print(f"Minimum Salary: {row['MinSalary']}")
        print("------")
```

```python
import csv

# Remote Jobs Filter
with open("job_posting.csv", "r") as f:
    reader = csv.DictReader(f)
    print("=== Remote Jobs ===")
    for row in reader:
        if row['Remote'] == 'Yes':
            print(f"{row['JobTitle']} — {row['Location']}")
```

```python
import csv

# High Salary Filter
with open("job_posting.csv", "r") as f:
    reader = csv.DictReader(f)
    print("=== High Salary Jobs ===")
    for row in reader:
        if int(row['MinSalary']) > 70000:
            print(f"{row['JobTitle']} — ${row['MinSalary']}")
```

```python
import csv

# Naya Row Add Karo
with open("job_posting.csv", "a") as f:
    writer = csv.writer(f)
    writer.writerow(["ML Engineer", "Bangalore", 90000, 130000, "Yes"])

print("New job added! ✅")

# Verify karo
with open("job_posting.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"{row['JobTitle']} — {row['Location']}")
```

```python
import csv
import os

# Remote + High Salary Filter
if os.path.exists("job_posting.csv"):
    with open("job_posting.csv", "r") as f:
        reader = csv.DictReader(f)
        print("=== Premium Remote Jobs ===")
        for row in reader:
            if row['Remote'] == 'Yes' and int(row['MinSalary']) > 80000:
                print(f"✅ {row['JobTitle']} — {row['Location']}")
                print(f"   Salary: ${row['MinSalary']} - ${row['MaxSalary']}")
else:
    print("File nahi mili! ❌")
```
