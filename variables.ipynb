```python
# Q10 — Candidate Score Card
c1_name = "Rahul"
c1_tech = 85
c1_comm = 70

c2_name = "Priya"
c2_tech = 60
c2_comm = 75

c3_name = "John"
c3_tech = 90
c3_comm = 88

c1_avg = (c1_tech + c1_comm) / 2
c2_avg = (c2_tech + c2_comm) / 2
c3_avg = (c3_tech + c3_comm) / 2

c1_status = "Shortlisted ✅" if c1_avg >= 70 else "Rejected ❌"
c2_status = "Shortlisted ✅" if c2_avg >= 70 else "Rejected ❌"
c3_status = "Shortlisted ✅" if c3_avg >= 70 else "Rejected ❌"

print(f"{c1_name} → Avg: {c1_avg} → {c1_status}")
print(f"{c2_name} → Avg: {c2_avg} → {c2_status}")
print(f"{c3_name} → Avg: {c3_avg} → {c3_status}")
```

```python
# Q11 — Smart Salary Negotiator
candidate_ask = 95000
company_budget = 85000

if candidate_ask <= company_budget:
    result = "Accepted as-is"
elif candidate_ask <= company_budget * 1.1:
    result = "Counter at budget"
elif candidate_ask <= company_budget * 1.2:
    result = "Negotiate"
else:
    result = "Too high — reject"

print(f"Candidate Ask  : ${candidate_ask}")
print(f"Company Budget : ${company_budget}")
print(f"Decision       : {result}")
```

```python
# Day 2 - Lists
favourite_food = ["Burger", "Chips", "Samosa", "Gujhiya", "Gulabjamun"]

print(favourite_food)
print(favourite_food[0])    # Pehla item
print(favourite_food[2])    # Teesra item
print(favourite_food[-1])   # Aakhri item
print(len(favourite_food))  # Total items
print("Burger" in favourite_food)  # True/False
```

```python
# List Methods
favourite_food.append("Pizza")
print(favourite_food)

favourite_food.remove("Samosa")
print(favourite_food)

favourite_food.insert(1, "Dosa")
print(favourite_food)

favourite_food.sort()
print(favourite_food)
```

```python
# Skills List — Recruiter Example
required_skills = ["Python", "SQL", "Excel", "Communication"]

required_skills.append("Machine Learning")
required_skills.remove("Excel")
required_skills.insert(1, "AI")
print("Python" in required_skills)
required_skills.sort()
print(required_skills)
print(len(required_skills))
```

```python
# Day 2 - Dictionary
myself = {
    "name"         : "Devansh",
    "age"          : 23,
    "city"         : "Kanpur",
    "current_role" : "AI Engineer",
    "salary"       : 62000
}

print(f"Name   : {myself['name']}")
print(f"Age    : {myself['age']}")
print(f"City   : {myself['city']}")
print(f"Role   : {myself['current_role']}")
print(f"Salary : {myself['salary']}")
```

```python
# Dictionary Methods
candidate = {
    "name"   : "Devansh",
    "age"    : 23,
    "city"   : "Kanpur",
    "salary" : 62000
}

print(candidate.get("name"))
candidate["goal"] = "AI Engineer"
candidate["city"] = "Mumbai"
print(candidate["city"])
print(candidate.keys())
print(candidate.values())
print(len(candidate))
print("salary" in candidate)
```
